```python
import pyttsx3    #allows Python to convert text into spoken audio.

engine = pyttsx3.init() #initialize the voices 

voices = engine.getProperty('voices')      #Retrieves a list of all available voices installed on your system (e.g., male and female voices).
male_voice = voices[0].id    # male voice (e.g., Microsoft David)
female_voice = voices[1].id  # female voice (e.g., Microsoft Zira)

def speak(text, voice_id):
    engine.setProperty('voice', voice_id)
    engine.say(text)
    engine.runAndWait()

# Male voice example
speak("Hello, this is the male voice speaking.", male_voice)

# Female voice example
speak("Hello, this is the female voice speaking.", female_voice)
```

```python
import speech_recognition as sr
from datetime import datetime

def save_note(text):
    timestamp = datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
    filename = f"note_{timestamp}.txt"
    with open(filename, "w", encoding='utf-8') as f:
        f.write(text)
    print(f"✅ Note saved as: {filename}")

def SpeachTOText():
    recognizer = sr.Recognizer()
    mic = sr.Microphone()

    print("🎙️ Speak your note. Press Ctrl+C to stop.")
    with mic as source:
        recognizer.adjust_for_ambient_noise(source)
        print("🔴 Listening...")
        audio = recognizer.listen(source)

    try:
        print("🧠 Transcribing...")
        text = recognizer.recognize_google(audio)
        print(f"📝 You said: {text}")
        save_note(text)
    except sr.UnknownValueError:
        print("❌ Could not understand your speech.")
    except sr.RequestError as e:
        print(f"🚫 Speech recognition service error: {e}")

if __name__ == "__main__":
    main()
```
